```python
import time
import numpy as np

from sklearn.datasets import fetch_covtype
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import SGDClassifier
from sklearn.ensemble import RandomForestClassifier, HistGradientBoostingClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

RANDOM_STATE = 42

print("Loading Forest Covertype dataset...")
data = fetch_covtype(as_frame=False)
X_full = data.data
y_full = data.target
print("Full dataset shape:", X_full.shape)

_, X, _, y = train_test_split(
    X_full, y_full,
    test_size=100_000,
    stratify=y_full,
    random_state=RANDOM_STATE
)

print("Stratified subsample shape:", X.shape)

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.25,
    stratify=y,
    random_state=RANDOM_STATE
)

print("Train size:", X_train.shape[0], "| Test size:", X_test.shape[0])

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

sgd_model = SGDClassifier(
    loss="log_loss",
    max_iter=1000,
    random_state=RANDOM_STATE,
    n_jobs=-1
)

rf_model = RandomForestClassifier(
    n_estimators=200,
    random_state=RANDOM_STATE,
    n_jobs=-1
)

hgb_model = HistGradientBoostingClassifier(
    random_state=RANDOM_STATE
)

print("\nTraining SGD Classifier...")
start = time.time()
sgd_model.fit(X_train_scaled, y_train)
sgd_train_time = time.time() - start

sgd_pred = sgd_model.predict(X_test_scaled)

sgd_accuracy = accuracy_score(y_test, sgd_pred)
sgd_precision = precision_score(
    y_test, sgd_pred, average="macro", zero_division=0
)
sgd_recall = recall_score(
    y_test, sgd_pred, average="macro", zero_division=0
)
sgd_f1 = f1_score(
    y_test, sgd_pred, average="macro", zero_division=0
)

print(
    "SGD Classifier -> Accuracy: %.4f | Precision: %.4f | Recall: %.4f | F1: %.4f"
    % (sgd_accuracy, sgd_precision, sgd_recall, sgd_f1)
)

print("\nTraining Random Forest...")
start = time.time()
rf_model.fit(X_train, y_train)
rf_train_time = time.time() - start

rf_pred = rf_model.predict(X_test)

rf_accuracy = accuracy_score(y_test, rf_pred)
rf_precision = precision_score(
    y_test, rf_pred, average="macro", zero_division=0
)
rf_recall = recall_score(
    y_test, rf_pred, average="macro", zero_division=0
)
rf_f1 = f1_score(
    y_test, rf_pred, average="macro", zero_division=0
)

print(
    "Random Forest -> Accuracy: %.4f | Precision: %.4f | Recall: %.4f | F1: %.4f"
    % (rf_accuracy, rf_precision, rf_recall, rf_f1)
)

print("\nTraining HistGradientBoosting...")
start = time.time()
hgb_model.fit(X_train, y_train)
hgb_train_time = time.time() - start

hgb_pred = hgb_model.predict(X_test)

hgb_accuracy = accuracy_score(y_test, hgb_pred)
hgb_precision = precision_score(
    y_test, hgb_pred, average="macro", zero_division=0
)
hgb_recall = recall_score(
    y_test, hgb_pred, average="macro", zero_division=0
)
hgb_f1 = f1_score(
    y_test, hgb_pred, average="macro", zero_division=0
)

print(
    "HistGradientBoosting -> Accuracy: %.4f | Precision: %.4f | Recall: %.4f | F1: %.4f"
    % (hgb_accuracy, hgb_precision, hgb_recall, hgb_f1)
)

model_names = [
    "SGD Classifier",
    "Random Forest",
    "HistGradientBoosting"
]

accuracies = [
    sgd_accuracy,
    rf_accuracy,
    hgb_accuracy
]

precisions = [
    sgd_precision,
    rf_precision,
    hgb_precision
]

recalls = [
    sgd_recall,
    rf_recall,
    hgb_recall
]

f1_scores = [
    sgd_f1,
    rf_f1,
    hgb_f1
]

train_times = [
    sgd_train_time,
    rf_train_time,
    hgb_train_time
]

print("\n================= FINAL COMPARISON =================")

print(
    "%-22s %-10s %-10s %-10s %-10s %-10s"
    % ("Model", "Accuracy", "Precision", "Recall", "F1-Score", "Time(s)")
)

for i in range(3):
    print(
        "%-22s %-10.4f %-10.4f %-10.4f %-10.4f %-10.2f"
        % (
            model_names[i],
            accuracies[i],
            precisions[i],
            recalls[i],
            f1_scores[i],
            train_times[i]
        )
    )

best_index = f1_scores.index(max(f1_scores))

print("\nBest performing algorithm: %s" % model_names[best_index])

print(
    "   F1-Score (macro) = %.4f | Accuracy = %.4f"
    % (
        f1_scores[best_index],
        accuracies[best_index]
    )
)
```
